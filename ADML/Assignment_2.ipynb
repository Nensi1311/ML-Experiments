### Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
```

```python
df_cleaned["Gender"] = encoder.fit_transform(df_cleaned["Gender"])
df_cleaned["Ever_Married"] = encoder.fit_transform(df_cleaned["Ever_Married"])
df_cleaned["Graduated"] = encoder.fit_transform(df_cleaned["Graduated"])
df_cleaned["Profession"] = encoder.fit_transform(df_cleaned["Profession"])
df_cleaned["Spending_Score"] = encoder.fit_transform(df_cleaned["Spending_Score"])
df_cleaned["Var_1"] = encoder.fit_transform(df_cleaned["Var_1"])
df_cleaned["Segmentation"] = encoder.fit_transform(df_cleaned["Segmentation"])
```

```python
df_cleaned.head()
```

### Standard Scaler

```python
from sklearn.preprocessing import StandardScaler
scaler= StandardScaler()
```

```python
scaler.fit_transform(df_cleaned)
```

### Data Split - train, test, valid

```python
from sklearn.model_selection import train_test_split
```

```python
X = df_cleaned.drop(columns=["Segmentation"])
Y = df_cleaned["Segmentation"]
```

```python
X_train, X_temp, Y_train, Y_temp = train_test_split(X, Y, train_size=0.7)
```

```python
X_valid, X_test, Y_valid, Y_test = train_test_split(X_temp, Y_temp, test_size=0.5)
```

# 2. Model Training
Train three classifiers: a Random Forest Classifier, an Extra Trees Classifier, and a Support Vector Machine (SVM) classifier on the preprocessed training data.

# 3. Model Evaluation
Evaluate the performance of the trained models (Random Forest, Extra Trees, and SVM) on the validation set. Report the accuracy scores for each classifier.

## 1. Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
```

```python
RF = RandomForestClassifier(n_estimators=100, random_state=42) 
```

```python
RF.fit(X_train, Y_train)
```

```python
Y_pred_rf = RF.predict(X_valid)
```

```python
print(classification_report(Y_valid, Y_pred_rf))
```

## 2. SVM

```python
from sklearn.svm import SVC 
```

```python
SVM = SVC(kernel="linear")
```

```python
SVM.fit(X_train, Y_train)
```

```python
Y_pred_svm = SVM.predict(X_valid)
```

```python
print(classification_report(Y_valid, Y_pred_svm))
```

## 3. Extra Tree Classifier

```python
from sklearn.ensemble import ExtraTreesClassifier
```

```python
ETC = ExtraTreesClassifier(n_estimators=100)
```

```python
ETC.fit(X_train, Y_train)
```

```python
Y_pred_etc = ETC.predict(X_valid)
```

```python
print(classification_report(Y_valid, Y_pred_etc))
```

## 4. Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingClassifier
```

```python
GBC = GradientBoostingClassifier(n_estimators=300,
                                 learning_rate=0.05,
                                 random_state=100,
                                 max_features=5)
```

```python
GBC.fit(X_train, Y_train)
```

```python
Y_pred_gbc = GBC.predict(X_valid)
```

```python
print(classification_report(Y_valid, Y_pred_gbc))
```

# 4. Ensemble Model
Combine the three classifiers into an Ensemble model using hard voting. Train the ensemble on the training data and evaluate it on the validation set. Does the ensemble model outperform the individual classifiers on the validation set? What are the advantages of using an ensemble model?

```python
from sklearn.ensemble import VotingClassifier
```

```python
ensemble_model = VotingClassifier(estimators=[
    ('RF', RF),
    ('ETC', ETC),
    ('SVM', SVM)
], voting='hard')
```

```python
ensemble_model.fit(X_train, Y_train)
```

```python
Y_pred_ensemble = ensemble_model.predict(X_valid)
```

```python
print(classification_report(Y_valid, Y_pred_ensemble))
```
